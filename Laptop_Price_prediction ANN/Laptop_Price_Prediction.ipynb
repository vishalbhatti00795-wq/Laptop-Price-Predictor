```python
df['Ram'] = df['Ram'].str.replace('GB', '').astype(int)
```

```python
df['Weight'] = df['Weight'].str.replace('kg', '').astype(float)
```

```python
df.head()
```

# One-**Hot** Encoding

```python
category = ['Company','TypeName','ScreenResolution','Cpu','Memory','Gpu','OpSys']
```

```python
df_encoded = pd.get_dummies(df, columns = category, drop_first=True)
```

```python
df_encoded
```

```python
df_encoded = df_encoded.drop(columns=['laptop_ID', 'Product'], errors='ignore')

df_encoded = df_encoded.astype(int)
```

```python
df_encoded
```

```python
df_encoded.head()
```

# Feature **Scaling**

```python
from sklearn.preprocessing import StandardScaler

numerical_features = ['Inches', 'Ram', 'Weight']

scaler = StandardScaler()

df_encoded[numerical_features] = scaler.fit_transform(df_encoded[numerical_features])
```

```python
from sklearn.model_selection import train_test_split

X = df_encoded.drop(columns = ['Price_euros'], axis=1)
y = df_encoded['Price_euros']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
X_train.shape
```

```python
y_train.shape
```

# Building an Artificial Neural Network(ANN)

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Dropout
from tensorflow.keras.callbacks import EarlyStopping
```

```python
model = Sequential()
```

```python
input_dim = X_train.shape[1]

model.add(Dense(units = 128, activation = 'relu', input_shape = (input_dim,)))
model.add(Dropout(0.2)) # First Hidden Layer

model.add(Dense(units = 128, activation = 'relu'))
model.add(Dropout(0.2)) # Second Hidden Layer

model.add(Dense(units = 64, activation = 'relu'))
model.add(Dropout(0.2)) # Third Hidden Layer

model.add(Dense(units = 32, activation = 'relu')) # Fourth Hidden Layer
model.add(Dropout(0.2))
```

```python
model.add(Dense(units = 1, activation = 'linear')) # Output Layer
```

```python
model.summary()
```

```python
early = EarlyStopping(monitor='val_loss', patience = 5, verbose = 1)
```

```python
model.compile(optimizer = 'adam', loss = 'mean_squared_error', metrics = ['mean_absolute_error'])
```

```python
history = model.fit(X_train, y_train, validation_split=0.2, epochs=100, batch_size = 32, callbacks=[early])
```

```python
y_pred = model.predict(X_test)
```

```python
import pickle

# Save the fitted StandardScaler
with open('scaler.pkl', 'wb') as f:
    pickle.dump(scaler, f)

# Save the trained Keras model
model.save('laptop_price_model.keras')

print("Successfully saved 'scaler.pkl' and 'laptop_price_model.keras'!")
```
